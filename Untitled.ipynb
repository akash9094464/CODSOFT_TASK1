```python
df_clean= df.copy()           #create a copy of dataset
```

```python
df_clean.isnull().sum()       #check missing values again
```

```python
df_clean["Age"].median()
```

```python
df_clean["Age"]= df_clean["Age"].fillna(df_clean["Age"].median())      #fill missing ages
```

```python
df_clean["Embarked"].mode()
```

```python
df_clean["Embarked"]= df_clean["Embarked"].fillna(df_clean["Embarked"].mode()[0])
```

```python
(df_clean["Cabin"].isnull().sum()/len(df_clean))*100
```

```python
df_clean.drop("Cabin", axis=1, inplace= True)
```

```python
df_clean.drop(
    ["PassengerId", "Name", "Ticket"],
    axis=1,
    inplace= True
)
```

```python
df_clean.columns
```

```python
df_clean["Sex"]= df_clean["Sex"].map({
    "male":0,
    "female":1
})
```

```python
df_clean.head()
```

```python
df_clean["Embarked"].unique()
```

```python
df_clean= pd.get_dummies(
    df_clean,
    columns=["Embarked"],
    drop_first=True
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean.dtypes
```

```python
x= df_clean.drop("Survived", axis=1)
y= df_clean["Survived"]
```

```python
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
lr = LogisticRegression(max_iter=1000)

lr.fit(x_train, y_train)
```

```python
y_pred_lr = lr.predict(x_test)
```

```python
lr_accuracy = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Accuracy:", lr_accuracy)
```

```python
cm_lr = confusion_matrix(y_test, y_pred_lr)

print(cm_lr)
```

```python
print(classification_report(
    y_test,
    y_pred_lr
))
```

```python
dt = DecisionTreeClassifier(
    random_state=42
)

dt.fit(x_train, y_train)
```

```python
y_pred_dt = dt.predict(x_test)
```

```python
dt_accuracy = accuracy_score(
    y_test,
    y_pred_dt
)

print(
    "Decision Tree Accuracy:",
    dt_accuracy
)
```

```python
print(
    classification_report(
        y_test,
        y_pred_dt
    )
)
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(x_train, y_train)
```

```python
y_pred_rf = rf.predict(x_test)
```

```python
rf_accuracy = accuracy_score(
    y_test,
    y_pred_rf
)

print(
    "Random Forest Accuracy:",
    rf_accuracy
)
```

```python
print(
    classification_report(
        y_test,
        y_pred_rf
    )
)
```

```python
results = {
    "Logistic Regression": lr_accuracy,
    "Decision Tree": dt_accuracy,
    "Random Forest": rf_accuracy
}

for model, score in results.items():
    print(
        f"{model}: {score:.4f}"
    )
```

```python
import matplotlib.pyplot as plt

models = list(results.keys())
scores = list(results.values())

plt.figure(figsize=(8,5))

plt.bar(models, scores)

plt.title(
    "Model Accuracy Comparison"
)

plt.ylabel("Accuracy")

plt.show()
```
